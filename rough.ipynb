```python
def merge(A,B):
    (m,n) = (len(A),len(B))
    (C,i,j,k) = ([],0,0,0)
    while k < m+n:
        if i == m:
            C.extend(B[j:])
            k = k + (n-j)
        elif j == n:
            C.extend(A[i:])
            k = k + (m-i)
        elif A[i] < B[j]:
            C.append(A[i])
            (i,k) = (i+1,k+1)
        else:
            C.append(B[j])
            (j,k) = (j+1,k+1)
    return(C)

def mergesort(A):
    n = len(A)
    if n <= 1:
         return(A)
    L = mergesort(A[:n//2])
    R = mergesort(A[n//2:])
    # return(L,R)
    B = merge(L,R)
    return(B)
```

```python
def quicksort(L,l,r): # Sort L[l:r]
    if (r - l <= 1):
        return(L)
    (pivot,lower,upper) = (L[l],l+1,l+1)
    for i in range(l+1,r):
        if L[i] > pivot: # Extend upper segment
            upper = upper+1
        else: # Exchange L[i] with start of upper segment
            (L[i], L[lower]) = (L[lower], L[i])
            # Shift both segments
            (lower,upper) = (lower+1,upper+1)
        # Move pivot between lower and upper
        (L[l],L[lower-1]) = (L[lower-1],L[l])
        lower = lower-1
    # Recursive calls
    quicksort(L,l,lower)
    quicksort(L,lower+1,upper)
    return(L)
```

```python
def quicksort(L, l, r):
    if r - l <= 1:
        return
    pivot, lower, upper = L[l], l + 1, l + 1
    for i in range(l + 1, r):
        if L[i] > pivot:
            upper += 1
        else:
            L[i], L[lower] = L[lower], L[i]
            lower += 1
            upper += 1
    L[l], L[lower - 1] = L[lower - 1], L[l]
    lower -= 1
    quicksort(L, l, lower)
    quicksort(L, lower + 1, r)
```

```python
def quicksort_mid(L, l, r):
    if r - l <= 1:
        return
    mid = (l + r) // 2
    L[l], L[mid] = L[mid], L[l]  # Swap middle to front
    pivot, lower, upper = L[l], l + 1, l + 1
    for i in range(l + 1, r):
        if L[i] > pivot:
            upper += 1
        else:
            L[i], L[lower] = L[lower], L[i]
            lower += 1
            upper += 1
    L[l], L[lower - 1] = L[lower - 1], L[l]
    lower -= 1
    quicksort_mid(L, l, lower)
    quicksort_mid(L, lower + 1, r)
```

```python
import time

class TimerError(Exception):
    """A custom exception used to report errors in use of Timer class"""

class Timer:
    def __init__(self):
        self._start_time = None
        self._elapsed_time = None

    def start(self):
        """Start a new timer"""
        if self._start_time is not None:
            raise TimerError("Timer is running. Use .stop()")
        self._start_time = time.perf_counter()

    def stop(self):
        """Save the elapsed time and re-initialize timer"""
        if self._start_time is None:
           raise TimerError("Timer is not running. Use .start()")
        self._elapsed_time = time.perf_counter() - self._start_time
        self._start_time = None

    def elapsed(self):
        """Report elapsed time"""
        if self._elapsed_time is None:
           raise TimerError("Timer has not been run yet. Use .start()")
        return(self._elapsed_time)

    def __str__(self):
        """print() prints elapsed time"""
        return(str(self._elapsed_time))
```

```python
import random

def compare_algorithms(n, Timer):
    """Compare regular quicksort and mid-pivot quicksort on sorted & unsorted inputs"""
    # Generate unsorted input
    unsorted_list = list(range(n))
    random.shuffle(unsorted_list)

    # Copy for mid-pivot quicksort
    list1 = unsorted_list.copy()
    list2 = unsorted_list.copy()

    timer = Timer()
    # Regular quicksort on unsorted input
    timer.start()
    quicksort(list1, 0, len(list1)-1)
    timer.stop()
    regular_unsorted_time = timer.elapsed()

    # Mid-pivot quicksort on unsorted input
    timer.start()
    quicksort_mid(list2, 0, len(list2)-1)
    timer.stop()
    mid_unsorted_time = timer.elapsed()

    # Generate sorted input
    sorted_list = list(range(n))
    list3 = sorted_list.copy()
    list4 = sorted_list.copy()

    # Regular quicksort on sorted input (worst case)
    timer.start()
    quicksort(list3, 0, len(list3)-1)
    timer.stop()
    regular_sorted_time = timer.elapsed()

    # Mid-pivot quicksort on sorted input (balanced)
    timer.start()
    quicksort_mid(list4, 0, len(list4)-1)
    timer.stop()
    mid_sorted_time = timer.elapsed()

    return {
        "regular_unsorted": regular_unsorted_time,
        "mid_unsorted": mid_unsorted_time,
        "regular_sorted": regular_sorted_time,
        "mid_sorted": mid_sorted_time
    }
```

```python
timer = Timer
n = 10
results = compare_algorithms(n, timer)
```
